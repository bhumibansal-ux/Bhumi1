3.opening a local csv file through url

```python
import requests
from io import StringIO

url = "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

pd.read_csv(data)
```

5.Index col parameter

```python
pd.read_csv('StudentPlacementData.csv',index_col='Branch')
```

6.header parameter

```python
pd.read_csv('StudentPlacementData.csv',header=1)
```

7.use_cols parameter

```python
pd.read_csv('StudentPlacementData.csv',usecols=['Gender','Branch','College_Tier','CGPA'])
```

8.squeeze parameter

```python
pd.read_csv('StudentPlacementData.csv',usecols=['Gender']).squeeze()
```

9.skiprows/nrows parameter

```python
pd.read_csv('StudentPlacementData.csv',nrows=100)
```

10.encoding parameter

11.skip bad lines

```python
pd.read_csv('BX-Books.csv', sep=';', encoding="latin-1", on_bad_lines='skip')
```

12.dtypes parameter

```python
pd.read_csv('StudentPlacementData.csv',dtype={'target':int}).info()
```

13.handling dates

```python
pd.read_csv('IPL Matches 2008-2020.csv',parse_dates=['date']).info()
```

```python
def rename(name):
    if name == "Royal Challengers Bangalore":
        return "RCB"
    else:
        return name
```

```python
rename("Royal Challengers Bangalore")
```

14.convertors

```python
pd.read_csv('IPL Matches 2008-2020.csv',converters={'team1':rename})
```

15.na_values parameter

```python
pd.read_csv('StudentPlacementData.csv',na_values=['Female',])
```

16.loading a huge dataset in chunks


```python
dfs = pd.read_csv('StudentPlacementData.csv',chunksize=5000)
for chunks in dfs:
    print(chunks.shape)
```
